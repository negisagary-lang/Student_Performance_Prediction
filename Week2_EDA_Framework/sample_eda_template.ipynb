## 2. Load Dataset

Replace the filename with the actual dataset path when available.

> **Note:** This is a placeholder. The real dataset is expected in a later phase.

```python
# df = pd.read_csv('students.csv')  # <- Uncomment when real data is available

# For demonstration, we create a small synthetic sample
# ------------------------------------------------
# ILLUSTRATIVE EXAMPLE ONLY - NOT REAL DATA
rng = np.random.default_rng(42)
df = pd.DataFrame({
    'Student_ID': range(1, 101),
    'Age': rng.integers(14, 18, 100),
    'Gender': rng.choice(['Male', 'Female'], 100),
    'Study_Hours': rng.normal(5, 2, 100).clip(0, 15),
    'Attendance': rng.normal(85, 10, 100).clip(0, 100),
    'Previous_Score': rng.normal(70, 15, 100).clip(0, 100),
    'Internet_Access': rng.choice(['Yes', 'No'], 100),
    'Final_Score': rng.normal(75, 12, 100).clip(0, 100),
})
df.loc[5, 'Study_Hours'] = np.nan  # introduce a missing value for demo
df.head()
```

## 3. Inspect Dataset

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
```

```python
df.info()
```

## 4. Data Quality Checks

### 4.1 Missing Values

```python
missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_report = pd.DataFrame({'Missing_Count': missing, 'Missing_Percent': missing_percent})
missing_report[missing_report['Missing_Count'] > 0]
```

### 4.2 Duplicate Records

```python
print('Duplicate rows:', df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
print('After dropping duplicates, shape:', df.shape)
```

## 5. Statistical Summary

```python
df.describe().T
```

## 6. Univariate Analysis

Analyzing one variable at a time.

### 6.1 Numerical Variable - Histogram + KDE

**Proposed plot:** Distribution of Final Score

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(df['Final_Score'], kde=True, bins=20, color='steelblue', ax=axes[0])
axes[0].set_title('Distribution of Final Score')
sns.kdeplot(df['Final_Score'], fill=True, color='seagreen', ax=axes[1])
axes[1].set_title('KDE of Final Score')
plt.tight_layout()
plt.show()
```

### 6.2 Numerical Variable - Box Plot (Outlier Detection)

**Proposed plot:** Box Plot of Study Hours

```python
plt.figure(figsize=(6, 5))
sns.boxplot(y=df['Study_Hours'], color='orange')
plt.title('Box Plot of Study Hours')
plt.show()
```

### 6.3 Categorical Variable - Bar Chart

**Proposed plot:** Gender distribution

```python
counts = df['Gender'].value_counts()
plt.figure(figsize=(6, 5))
sns.barplot(x=counts.index, y=counts.values, palette='viridis')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

## 7. Bivariate Analysis

Analyzing relationships between two variables.

### 7.1 Numerical + Numerical - Scatter Plot

**Proposed plot:** Study Hours vs Final Score

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Study_Hours', y='Final_Score', alpha=0.7)
plt.title('Study Hours vs Final Score')
plt.show()
```

### 7.2 Categorical + Numerical - Box Plot

**Proposed plot:** Internet Access vs Final Score

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Internet_Access', y='Final_Score', palette='Set2')
plt.title('Internet Access vs Final Score')
plt.show()
```

### 7.3 Categorical + Categorical - Crosstab / Grouped Bar

**Proposed plot:** Gender vs Internet Access

```python
ct = pd.crosstab(df['Gender'], df['Internet_Access'])
ct.plot(kind='bar', figsize=(7, 5), color=['#5b9bd5', '#ed7d31'])
plt.title('Gender vs Internet Access (Crosstab)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
print(ct)
```

## 8. Multivariate Analysis

Analyzing interactions among multiple variables.

### 8.1 Correlation Matrix + Heatmap

**Proposed plot:** Correlation heatmap of numerical features

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
corr = df[num_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```
